Simple Calculator Class

```python
class Calculator:
    def add(self, a, b):
        return a + b
    
    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        return a / b
calc = Calculator()
print(calc.add(5, 3))        
print(calc.subtract(5, 3))   
print(calc.multiply(5, 3))   
print(calc.divide(10, 2))    
    
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        if 0 <= grade <= 100:
            self.grades.append(grade)
        else:
            print("Invalid grade ,Grade must be between 0 and 100")

    def calculate_average(self):
        return sum(self.grades) / len(self.grades)

    def grade(self):
        avg = self.calculate_average()
        if avg >= 90:
            return "A"
        elif avg >= 80:
            return "B"
        elif avg >= 70:
            return "C"
        elif avg >= 60:
            return "D"
        else:
            return "F"

    def Student_details(self):
        print(f"Student: {self.name}")
        print(f"Grades: {self.grades}")
        print(f"Average: {self.calculate_average()}")
        print(f"Final Grade: {self.grade()}")

student1 = Student("Zyad")

student1.add_grade(95)
student1.add_grade(82)
student1.add_grade(76)
student1.add_grade(60)
student1.Student_details()

```

Modeling a Bank Account

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount} and New balance: {self.balance}")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Withdrawal failed")
        elif amount <= 0:
            print("Withdraw amount must be positive")
        else:
            self.balance -= amount
            print(f"Withdrew {amount} and New balance: {self.balance}")

    def show_balance(self):
        print(f"Your Balance: {self.balance}")


account1 = BankAccount(1000)

account1.deposit(500)
account1.withdraw(200)
account1.withdraw(1500)  
account1.show_balance()
```

Modeling a Simple Library System

```python
class Library:
    def __init__(self):
        self.books = []  
        self.borrowed = {}  

    def add_book(self, book_name):
        self.books.append(book_name)

    def show_books(self):
        if self.books:
            print("Available books in library:")
            for book in self.books:
                print(f"{book}")
        else:
            print("No books available in the library")

    def borrow_book(self, book_name, borrower_name):
        if book_name in self.books:
            self.books.remove(book_name)
            self.borrowed[book_name] = borrower_name
            print(f'Book "{book_name}" borrowed by {borrower_name}')
        else:
            print(f'Sorry, "{book_name}" is not available')


library = Library()

library.add_book("Python Programming")
library.add_book("Data Science")
library.add_book("AI for Beginners")
library.show_books()
library.borrow_book("Python Programming", "Zyad")
library.show_books()


```

Simple Voting System

```python
class VotingSystem:
    def __init__(self):
        self.candidates = {}  

    def add_candidate(self, name):
        if name not in self.candidates:
            self.candidates[name] = 0
            print(f"Candidate '{name}' added")
        else:
            print(f"Candidate '{name}' already exists")

    def vote(self, name):
        if name in self.candidates:
            self.candidates[name] += 1
            print(f"Vote casted for {name}")
        else:
            print(f"Candidate '{name}' does not exist")

    def show_results(self):
        print("Voting Results:")
        if not self.candidates:
            print("No candidates available")
            return
        for name, votes in self.candidates.items():
            print(f"{name}: {votes} vote")

    def winner(self):
        if not self.candidates:
            print("No candidates available")
            return
        max_votes = max(self.candidates.values())
        winners = [name for name, votes in self.candidates.items() if votes == max_votes]
        print(f"Winner: {', '.join(winners)} with {max_votes} vote")

voting = VotingSystem()
voting.add_candidate("Zyad")
voting.add_candidate("Ahmed")
voting.add_candidate("Atef")
voting.vote("Zyad")
voting.vote("Ahmed")
voting.vote("Zyad")
voting.vote("Atef")
voting.vote("Mohamed")
voting.show_results()
voting.winner()
```

Simple Temperature Converter

```python
class TemperatureConverter:
    def __init__(self, temperature=0):
        self.temperature = temperature

    def to_fahrenheit(self):
        return (self.temperature * 9/5) + 32

    def to_kelvin(self):
        return self.temperature + 273.15
    
    def from_fahrenheit(self):
        return (self.temperature - 32) * 5/9

    def from_kelvin(self):
        return self.temperature - 273.15
    

temp_c = TemperatureConverter(25)
print(f"25°C in Fahrenheit: {temp_c.to_fahrenheit()}°F")
print(f"25°C in Kelvin: {temp_c.to_kelvin()}K")

temp_f = TemperatureConverter(77)
print(f"77°F in Celsius: {temp_f.from_fahrenheit()}°C")

temp_k = TemperatureConverter(300)
print(f"300K in Celsius: {temp_k.from_kelvin():.2f}°C")
```

Modeling a Car Rental System

```python
class CarRental:
    def __init__(self):
        self.available_cars = []  
        self.rented_cars = {}     

    def add_car(self, car_name):
            self.available_cars.append(car_name)
            print(f'Car "{car_name}" added to the fleet')


    def show_available_cars(self):
        if self.available_cars:
            print("Available Cars:")
            for car in self.available_cars:
                print(f"{car}")
        else:
            print("No cars available for rent")

    def rent_car(self, car_name, renter_name):
        if car_name in self.available_cars:
            self.available_cars.remove(car_name)
            self.rented_cars[car_name] = renter_name
            print(f'Car "{car_name}" rented by {renter_name}')
        else:
            print(f'Sorry, "{car_name}" is not available for rent')

    def return_car(self, car_name):
        if car_name in self.rented_cars:
            renter = self.rented_cars.pop(car_name)
            self.available_cars.append(car_name)
            print(f'Car "{car_name}" returned by {renter}')
        else:
            print(f'Car "{car_name}" was not rented')

rental = CarRental()

rental.add_car("Toyota")
rental.add_car("Honda")
rental.add_car("BMW")
rental.show_available_cars()
rental.rent_car("Toyota", "Zyad")
rental.show_available_cars()
rental.return_car("Toyota")
rental.show_available_cars()
```
